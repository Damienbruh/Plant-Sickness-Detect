# Plant Disease Detection - Data Exploration

This notebook explores the plant disease dataset to understand the data distribution and characteristics.

## 1. Imports

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
from pathlib import Path

# Set visualization style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 8)
```

## 2. Dataset Structure and Distribution

```python
# Define dataset paths
BASE_DIR = Path('..') / 'Dataset'
TRAIN_DIR = BASE_DIR / 'Train' / 'Train'
VAL_DIR = BASE_DIR / 'Validation' / 'Validation'
TEST_DIR = BASE_DIR / 'Test' / 'Test'

# Get class names
classes = sorted([d.name for d in TRAIN_DIR.iterdir() if d.is_dir()])
print(f'Classes found: {classes}')
print(f'Number of classes: {len(classes)}')
```

```python
# Count images in each split
def count_images(directory):
    counts = {}
    for class_name in classes:
        class_path = directory / class_name
        if class_path.exists():
            counts[class_name] = len(list(class_path.glob('*')))
        else:
            counts[class_name] = 0
    return counts

train_counts = count_images(TRAIN_DIR)
val_counts = count_images(VAL_DIR)
test_counts = count_images(TEST_DIR)

# Create DataFrame for visualization
df_counts = pd.DataFrame({
    'Train': train_counts,
    'Validation': val_counts,
    'Test': test_counts
})

print('\nDataset Distribution:')
print(df_counts)
print(f'\nTotal images: {df_counts.sum().sum()}')
```

```python
# Visualize dataset distribution
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Plot 1: Distribution per split
df_counts.plot(kind='bar', ax=axes[0])
axes[0].set_title('Image Distribution Across Splits')
axes[0].set_xlabel('Class')
axes[0].set_ylabel('Number of Images')
axes[0].legend(title='Split')
axes[0].tick_params(axis='x', rotation=45)

# Plot 2: Total distribution per class
total_per_class = df_counts.sum(axis=1)
axes[1].bar(total_per_class.index, total_per_class.values, color='skyblue')
axes[1].set_title('Total Images Per Class')
axes[1].set_xlabel('Class')
axes[1].set_ylabel('Total Number of Images')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

## 3. Sample Images from Each Class

```python
# Display sample images
num_samples = 3
fig, axes = plt.subplots(len(classes), num_samples, figsize=(12, 4*len(classes)))

for i, class_name in enumerate(classes):
    class_path = TRAIN_DIR / class_name
    images = list(class_path.glob('*.jpg')) + list(class_path.glob('*.png'))
    
    for j in range(min(num_samples, len(images))):
        img = Image.open(images[j])
        axes[i, j].imshow(img)
        axes[i, j].axis('off')
        if j == 0:
            axes[i, j].set_title(f'{class_name}\nSize: {img.size}', fontsize=12)
        else:
            axes[i, j].set_title(f'Size: {img.size}', fontsize=10)

plt.tight_layout()
plt.show()
```

## 4. Image Properties Analysis

```python
# Analyze image properties (size, format, etc.)
image_info = []

for class_name in classes:
    class_path = TRAIN_DIR / class_name
    images = list(class_path.glob('*.jpg')) + list(class_path.glob('*.png'))
    
    for img_path in images[:50]:  # Sample first 50 images per class
        img = Image.open(img_path)
        image_info.append({
            'class': class_name,
            'width': img.width,
            'height': img.height,
            'format': img.format,
            'mode': img.mode
        })

df_images = pd.DataFrame(image_info)

print('Image Properties Summary:')
print(df_images.groupby('class')[['width', 'height']].describe())
print(f'\nImage formats: {df_images["format"].unique()}')
print(f'Image modes: {df_images["mode"].unique()}')
```

```python
# Visualize image dimensions
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Width distribution
for class_name in classes:
    class_data = df_images[df_images['class'] == class_name]
    axes[0].hist(class_data['width'], alpha=0.5, label=class_name, bins=20)
axes[0].set_title('Image Width Distribution')
axes[0].set_xlabel('Width (pixels)')
axes[0].set_ylabel('Frequency')
axes[0].legend()

# Height distribution
for class_name in classes:
    class_data = df_images[df_images['class'] == class_name]
    axes[1].hist(class_data['height'], alpha=0.5, label=class_name, bins=20)
axes[1].set_title('Image Height Distribution')
axes[1].set_xlabel('Height (pixels)')
axes[1].set_ylabel('Frequency')
axes[1].legend()

plt.tight_layout()
plt.show()
```
